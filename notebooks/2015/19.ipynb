## Part a

### Brute-force approach
Let's start by simply brute-forcing all possible replacements

```python
# Imports
import re
```

```python
# Functions
def solve_a_brute_force(input_data: str) -> int:
    """Brute-force approach to generate all possible molecules from one replacement."""
    # Parse the input into replacements and the initial molecule
    replacements_str, molecule = input_data.split("\n\n")
    replacements: list[list[str]] = [line.split(" => ") for line in replacements_str.splitlines()]

    # Generate all possible molecules by applying each replacement to each occurrence in the molecule
    new_molecules: set[str] = {
        # Replace the matched substring with the replacement string
        molecule[: match.start()] + to_str + molecule[match.end() :]
        # For each replacement and each match of the "from" string in the molecule
        for from_str, to_str in replacements
        for match in re.finditer(from_str, molecule)
    }

    # Return the number of unique molecules generated
    return len(new_molecules)
```

```python
# Correctness check
check_example(solve_a_brute_force, example)
```

```python
# Performance check
time_a = time_solution(solve_a_brute_force, input_data)
```

The brute force solution takes about 0.36 ms, which is fast enough for now.

```python
# Submit answer
puzzle.answer_a = solve_a_brute_force(input_data)
```
